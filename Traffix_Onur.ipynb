```python
import kagglehub
import os
import shutil

print("1. Veri seti Kaggle'dan indiriliyor (Bu işlem 7GB için biraz sürebilir)...")
# Veri seti gizli bir cache klasörüne iner
dataset_path = kagglehub.dataset_download('omertarikyilmaz/istabul-traffic-2020-2024')
print(f"-> Veri indirme tamamlandı. Gizli konum: {dataset_path}")

print("\n2. Dosya Colab ana çalışma dizinine taşınıyor...")
# Senin veri_on_isleme.py dosyanın beklediği tam isim ve konum
hedef_dosya_yolu = '/content/drive/MyDrive/Traffix/istanbulTraffic2020-2024.csv'

# İndirilen klasördeki CSV dosyasını dinamik olarak bulalım (Kaggle'daki orijinal adı ne olursa olsun)
csv_dosyalari = [f for f in os.listdir(dataset_path) if f.endswith('.csv')]

if csv_dosyalari:
    orijinal_dosya_yolu = os.path.join(dataset_path, csv_dosyalari[0])

    # 7GB veriyi kopyalamak disk alanını ikiye katlar, bu yüzden 'taşıma' (move) işlemi yapıyoruz
    shutil.move(orijinal_dosya_yolu, hedef_dosya_yolu)
    print(f"✅ BAŞARILI: Veri seti '{hedef_dosya_yolu}' olarak ana dizine yerleştirildi!")
    print("Artık veri_on_isleme.py dosyanı hiçbir değişiklik yapmadan çalıştırabilirsin.")
else:
    print("❌ HATA: İndirilen klasörde herhangi bir CSV dosyası bulunamadı.")
```
